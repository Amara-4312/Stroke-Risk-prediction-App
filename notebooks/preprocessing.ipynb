```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/healthcare-dataset-stroke-data.csv')

# Drop id column - it's just a row identifier, no predictive value
df = df.drop(columns=['id'])

# Check the Other gender rows
print("Gender value counts:\n", df['gender'].value_counts())
print("\nShape after dropping id:", df.shape)
```

```python
# Drop the single 'Other' gender row
df = df[df['gender'] != 'Other']

print("Shape after dropping Other gender:", df.shape)
print("\nGender value counts:\n", df['gender'].value_counts())
```

```python
# Impute missing BMI with median
# We use median instead of mean because BMI has some high outliers
# that would skew the mean upwards

bmi_median = df['bmi'].median()
print("BMI median:", bmi_median)

df['bmi'] = df['bmi'].fillna(bmi_median)

# Confirm no more missing values
print("\nMissing values after imputation:")
print(df.isnull().sum())
```

```python
# Encode categorical columns
# Binary columns (only 2 categories) - simple map
df['gender'] = df['gender'].map({'Female': 0, 'Male': 1})
df['ever_married'] = df['ever_married'].map({'No': 0, 'Yes': 1})
df['Residence_type'] = df['Residence_type'].map({'Rural': 0, 'Urban': 1})

# Multi-category columns - one hot encoding
df = pd.get_dummies(df, columns=['work_type', 'smoking_status'], drop_first=False)

print("Shape after encoding:", df.shape)
print("\nColumns:\n", df.columns.tolist())
print("\nFirst row:\n", df.iloc[0])
```

```python
# Convert True/False to 1/0
df = df.astype({col: int for col in df.select_dtypes(bool).columns})

# Confirm
print("Data types:\n", df.dtypes)
print("\nFirst row:\n", df.iloc[0])

# Save the cleaned dataset
df.to_csv('../data/stroke_cleaned.csv', index=False)
print("\n✅ Cleaned dataset saved to data/stroke_cleaned.csv")
```
